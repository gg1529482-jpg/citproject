```python
# ============================================================
# TEXT-TO-IMAGE GENERATION PROJECT
# GAN-BASED GENERATIVE AI DEMONSTRATION
# Google Colab - No files or folders required
# ============================================================

# 1. Install required libraries
!pip -q install diffusers transformers accelerate

# 2. Import libraries
import torch
import matplotlib.pyplot as plt
from diffusers import DiffusionPipeline

# 3. Select GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)

# 4. Load pretrained text-to-image model
pipe = DiffusionPipeline.from_pretrained(
    "stabilityai/sd-turbo",
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
)

pipe = pipe.to(device)

# 5. Give your text prompt
prompt = "a beautiful futuristic city with trees and mountains, digital art"

print("Text Prompt:", prompt)
print("Generating image...")

# 6. Generate image
image = pipe(
    prompt=prompt,
    num_inference_steps=1,
    guidance_scale=0.0
).images[0]

# 7. Display generated image
plt.figure(figsize=(8, 8))
plt.imshow(image)
plt.axis("off")
plt.title("TEXT-TO-IMAGE GENERATED OUTPUT")
plt.show()

print("Image generation completed successfully!")
```

```python
import os
from google import genai
from google.genai import types
from google.colab import userdata

api_key = userdata.get('dee17')

os.environ['dee17'] = api_key

client = genai.Client(api_key=api_key)
persona = """
you are a highly sarcastic, brutally honest.Senior python developer you find begginer questions annoying but you will ultimatly answer them you must always include
atleast one sarcastic remark before giving the correct code. Never apologize.keep your answer under 3 sentences."""

chat_config = types.GenerateContentConfig(
    system_instruction=persona,
    temperature=0.8
)

chat = client.chats.create(
    model='gemini-3.8-flash',
    config=chat_config
)

print("The Sarcastic Senior Dev is online. Ask a coding question! (Type 'exit' to stop)")
print("-" * 50)

while True:
    user_question = input("You: ")

    if user_question.lower() == "exit":
        print("Closing the terminal...")
        break

    response = chat.send_message(user_question)
    print(f"Sarcastic Senior Dev: {response.text}\n")
```
